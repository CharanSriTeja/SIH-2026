```python
import rasterio
from rasterio.windows import from_bounds
import numpy as np
import pandas as pd
from pyproj import Transformer

filepath = "smap_1km_data/your_downloaded_file.tif"

# your district's lat/lon bounding box
min_lon, min_lat, max_lon, max_lat = 91.5, 25.0, 92.5, 26.0

with rasterio.open(filepath) as src:
    # reproject the lat/lon box into the file's CRS
    transformer = Transformer.from_crs("EPSG:4326", src.crs, always_xy=True)
    x_min, y_min = transformer.transform(min_lon, min_lat)
    x_max, y_max = transformer.transform(max_lon, max_lat)

    # build a window from those projected bounds
    window = from_bounds(x_min, y_min, x_max, y_max, transform=src.transform)
    band1 = src.read(1, window=window)
    band1 = np.where(band1 == src.nodata, np.nan, band1)
    window_transform = src.window_transform(window)

print("Cropped shape:", band1.shape)   # should now be small — a few hundred x few hundred, not millions

# flatten just this small window into a table
rows, cols = np.indices(band1.shape)
xs, ys = rasterio.transform.xy(window_transform, rows.flatten(), cols.flatten())
xs, ys = np.array(xs), np.array(ys)

transformer_back = Transformer.from_crs(src.crs, "EPSG:4326", always_xy=True)
lons, lats = transformer_back.transform(xs, ys)

df = pd.DataFrame({
    "lat": lats,
    "lon": lons,
    "soil_moisture": band1.flatten()
})

df = df.dropna(subset=["soil_moisture"])   # drop empty pixels
print(df.shape)
print(df.head())

df.to_csv("district_soil_moisture.csv", index=False)
print("Saved to district_soil_moisture.csv")
```

```python
import rasterio
from rasterio.windows import from_bounds
from pyproj import Transformer
import numpy as np
import pandas as pd
import glob
import os

# ---- 1. Define your 8 districts with their bounding boxes ----
# Replace these with your actual district names and (min_lon, min_lat, max_lon, max_lat)
districts = {
    "District_1": (91.5, 25.0, 92.0, 25.5),
    "District_2": (92.0, 25.0, 92.5, 25.5),
    "District_3": (91.5, 25.5, 92.0, 26.0),
    "District_4": (92.0, 25.5, 92.5, 26.0),
    "District_5": (91.0, 25.0, 91.5, 25.5),
    "District_6": (91.0, 25.5, 91.5, 26.0),
    "District_7": (92.5, 25.0, 93.0, 25.5),
    "District_8": (92.5, 25.5, 93.0, 26.0),
}

def crop_to_bbox(src, band1, min_lon, min_lat, max_lon, max_lat):
    transformer = Transformer.from_crs("EPSG:4326", src.crs, always_xy=True)
    x_min, y_min = transformer.transform(min_lon, min_lat)
    x_max, y_max = transformer.transform(max_lon, max_lat)
    window = from_bounds(x_min, y_min, x_max, y_max, transform=src.transform)

    row_start, row_stop = int(window.row_off), int(window.row_off + window.height)
    col_start, col_stop = int(window.col_off), int(window.col_off + window.width)
    cropped = band1[row_start:row_stop, col_start:col_stop]
    cropped_transform = src.window_transform(window)
    return cropped, cropped_transform

# ---- 2. Loop through every downloaded .tif file ----
all_records = []

tif_files = sorted(glob.glob("smap_1km_data/*.tif"))
print(f"Processing {len(tif_files)} files...")

for filepath in tif_files:
    filename = os.path.basename(filepath)
    with rasterio.open(filepath) as src:
        band1 = src.read(1)
        band1 = np.where(band1 == src.nodata, np.nan, band1)

        for district_name, (min_lon, min_lat, max_lon, max_lat) in districts.items():
            cropped, cropped_transform = crop_to_bbox(src, band1, min_lon, min_lat, max_lon, max_lat)

            if cropped.size == 0:
                continue

            rows, cols = np.indices(cropped.shape)
            xs, ys = rasterio.transform.xy(cropped_transform, rows.flatten(), cols.flatten())
            xs, ys = np.array(xs), np.array(ys)

            transformer_back = Transformer.from_crs(src.crs, "EPSG:4326", always_xy=True)
            lons, lats = transformer_back.transform(xs, ys)

            values = cropped.flatten()

            for lat, lon, val in zip(lats, lons, values):
                if not np.isnan(val):
                    all_records.append({
                        "file": filename,
                        "district": district_name,
                        "lat": lat,
                        "lon": lon,
                        "soil_moisture": val
                    })

# ---- 3. Build final table ----
df = pd.DataFrame(all_records)
print(df.shape)
print(df.head())

df.to_csv("8_districts_soil_moisture.csv", index=False)
print("Saved to 8_districts_soil_moisture.csv")
```

```python
import pandas as pd
import camelot

# ----------------------------
# Step 1: Extract tables from PDF
# ----------------------------
# Install if needed:
# pip install camelot-py[cv]

pdf_path = "landslide_report.pdf"  # change to your file name/path

# 'lattice' works well for grid-like tables; try 'stream' if that fails
tables = camelot.read_pdf(pdf_path, pages="all", flavor="lattice")

# Combine all tables into one DataFrame
df_list = [t.df for t in tables]
df = pd.concat(df_list, ignore_index=True)

# ----------------------------
# Step 2: Clean and assign column names
# ----------------------------
expected_columns = [
    "Sl.No.", "Slide_No", "State", "District", "Slide_Name",
    "NH_SH_Location", "Latitude", "Longitude", "Material_Involved",
    "Movement_Type", "History"
]

# If the first row looks like a header, drop it
first_row = df.iloc[0].astype(str).str.strip()
if "State" in first_row.values or "Slide_No" in first_row.values:
    df = df[1:].reset_index(drop=True)

# Assign column names (ensure we have at least 11 columns)
df = df.iloc[:, :11]  # keep first 11 columns if there are extras
df.columns = expected_columns

# Optional: basic cleaning
for col in df.columns:
    df[col] = df[col].astype(str).str.strip()

# ----------------------------
# Step 3: Filter for the 8 states
# ----------------------------
target_states = [
    "Assam",
    "Arunachal Pradesh",
    "Himachal Pradesh",
    "Manipur",
    "Mizoram",
    "Nagaland",
    "Sikkim",
    "West Bengal"
]

# Normalize case just in case
df["State_clean"] = df["State"].str.title()
df_filtered = df[df["State_clean"].isin(target_states)].copy()
df_filtered = df_filtered.drop(columns=["State_clean"])

# ----------------------------
# Step 4: Save to CSV
# ----------------------------
output_csv = "north_east_landslides_filtered.csv"
df_filtered.to_csv(output_csv, index=False)

print(f"Total rows extracted: {len(df)}")
print(f"Rows after filtering (8 states): {len(df_filtered)}")
print(f"Saved to: {output_csv}")
```

```python
# Install needed libraries first (run once in your environment):
# pip install rasterio matplotlib

import rasterio
from rasterio.plot import show
import matplotlib.pyplot as plt
import numpy as np

# Path to one CartoDEM tile (change this to your actual file path)
dem_tile = "C1_DEM_16B_2005-2014_v3_R-1_88E21N_f45q\cdnf45q_v3r1\cdnf45q.tif"  # example filename

# Open the raster
with rasterio.open(dem_tile) as src:
    print("CRS:", src.crs)
    print("Resolution (m):", src.res)
    print("Shape (rows, cols):", src.shape)
    print("Bounds:", src.bounds)
    
    # Read elevation band
    elev = src.read(1)
    
    # Basic stats
    print("Min elevation (m):", np.nanmin(elev))
    print("Max elevation (m):", np.nanmax(elev))
    print("Mean elevation (m):", np.nanmean(elev))

    # Plot
    plt.figure(figsize=(8, 6))
    show(elev, cmap="terrain", title="CartoDEM Tile: " + dem_tile)
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.tight_layout()
    plt.show()
```
